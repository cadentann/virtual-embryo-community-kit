# VEC Data Inspector — synthetic demo

This notebook creates synthetic `.h5ad` files and writes a read-only inspection report to a separate directory. It does not use challenge data.

## Supported local environment

From the repository root, run `cd modules/data_inspector`, install with `python -m pip install -e '.[notebook]'`, and start `python -m jupyter lab notebooks/vec_data_inspector_demo.ipynb`. This release supports local Jupyter only; Google Colab is not assessed or supported.


```python
from pathlib import Path
from datetime import datetime, timezone
import subprocess, sys
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
run_dir = root / 'reports' / ('notebook-demo-' + datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ'))
fixture_dir = run_dir / 'fixtures'
report_dir = run_dir / 'inspection'
subprocess.run([sys.executable, 'examples/make_synthetic_fixtures.py', '--output-dir', str(fixture_dir)], cwd=root, check=True)
```

```python
command = [sys.executable, '-m', 'vec_data_inspector.cli',
           str(fixture_dir / 'synthetic_dense_t1_like.h5ad'),
           '--panel', str(fixture_dir / 'synthetic_panel_reordered.txt'),
           '--report-dir', str(report_dir), '--write-reorder-plan']
subprocess.run(command, cwd=root, check=True)
```

```python
from IPython.display import Markdown, display
display(Markdown((report_dir / 'inspection_report.md').read_text(encoding='utf-8')))
```

The CSV is a source-index plan only: neither this notebook nor the command changes the input file. A panel comparison is a structural signal, not a submission or authorization decision. Finite/non-negative values do not prove the required normalization, and `n_obs` is not embryo size. Reports made from real challenge files may contain unpublished or local identifying details; keep them private unless current terms permit the intended sharing, and use synthetic fixtures for public examples.
